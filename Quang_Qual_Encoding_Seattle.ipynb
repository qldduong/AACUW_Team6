```python
# Split Seattle 2021 DFs by year 
print(seattle_df["seattle_rescue_plan"].value_counts())

# Most programs fall under SRP1 

# Method for finding with program IDs correspond to a particular SRP (1 - 4). 
# Parameter: the SRP plan number you're lookign for (ex: 'SRP1', 'SRP2'...)
def ids_by_year(year_string):
    srp_ids = seattle_df.loc[seattle_df['seattle_rescue_plan'] == year_string, 'program_id']
    sorted_ids = srp_ids.sort_values()
    srp_id_list = sorted_ids.tolist()
    return(srp_id_list)


srp1_ids = ids_by_year('SRP1')
srp2_ids = ids_by_year('SRP2')
srp3_ids = ids_by_year('SRP3')
srp4_ids = ids_by_year('SRP4')  

print(srp1_ids)
print(srp2_ids)
print(srp3_ids)
print(srp4_ids)
```

## Note on qualitative evaluation 
Only a small subset of the programs in the dataset were actually reported with enough clarity + detail to be included in this part of analysis


```python
# df['category of investment']

# print(df['total spent'].agg))

# sorted_seattle = sorted_seattle[['program_id', 'public_name', 'program_status', 'budgeted', 'total_spent', 'seattle_rescue_plan']]

eval_ag_df = clean_eval_df.groupby('benef_type').agg(
    budget_sum = ("budgeted", "sum"),
    budget_avg = ("budgeted", "mean"), 
    spending_sum = ("total_spent", "sum"),
    spending_avg = ("total_spent", "mean"),
    program_count = ("program_id", "count")
)

# Convert sum and average to millions
eval_ag_df['budget_sum'] = eval_ag_df['budget_sum'] / 1000000
eval_ag_df['budget_avg'] = eval_ag_df['budget_avg'] / 1000000
eval_ag_df['spending_sum'] = eval_ag_df['spending_sum'] / 1000000
eval_ag_df['spending_avg'] = eval_ag_df['spending_avg'] / 1000000

print(eval_ag_df)
print(f"Total budgeted: {sum(eval_ag_df["budget_sum"])}")
print(f"Total spent: {sum(eval_ag_df["spending_sum"])}")
```

## Making Visualizations for Spending Based on benef_type

```python
## Bar graph comparing the total budgeted and the total spent by beneficiary type

# Set the width of the bars
bar_width = 0.35

# Create a figure and axis
fig, ax = plt.subplots(figsize=(10, 6))

# Plot both the budget_sum and total_spent_sum as bars
index = np.arange(len(eval_ag_df))  # Position of the bars on the x-axis
ax.bar(index, eval_ag_df['budget_sum'], bar_width, label='Budget Sum', color='b')
ax.bar(index + bar_width, eval_ag_df['spending_sum'], bar_width, label='Total Spent Sum', color='r')

# Set x-tick labels as the category names
ax.set_xticks(index + bar_width / 2)
ax.set_xticklabels(eval_ag_df.index, rotation=45)

# Adding labels and title
ax.set_xlabel('Beneficiary Type')
ax.set_ylabel('Amount (Millions of Dollars)')
ax.set_title('Comparison of Total Spent and Budget by Beneficiary Type')

# Adding legend
ax.legend()

# Show the plot
plt.show()
```
